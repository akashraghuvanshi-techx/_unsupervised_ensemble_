**Feature Engineering**

Feature 1 . Overall Skill Score

We create an overall skill score by taking the average of crossing, finishing, dribbling, ball control and short passing. This gives a combined measure of a players overall technical and attacking skills.

```python
df["overall_skill_score"] = df[["crossing","finishing","dribbling","ball_control","short_passing"]].mean(axis=1)
df[["overall_skill_score"]].head()
```

 Feature 2 Defensive Skill Score

We create a defensive skill score by taking the average of marking, standing tackle and interceptions. This gives a combined measure of a player overall defensive abilities.

```python
df["defensive_skill_score"] = df[["marking","standing_tackle","interceptions"]].mean(axis=1)
df[["defensive_skill_score"]].head()
```

# ** Ensemble Learning**

In this section.... we will use ensemble learning techniques to predict the players overall rating. We will start with Random Forest.

Our target variable is overall_rating. It represents the overall rating of a football player which we want to predict using the player other attributes.

```python
y = df["overall_rating"]
X = df.drop(columns = ["overall_rating"])
```

```python
X = X.drop(columns = ["id","player_fifa_api_id","player_api_id","date"])
```

```python
X.select_dtypes(include = "object").columns
```

 Encoding Categorical Features

The categorical features cannot be directly used by the Random Forest model. we will convert them into numerical form using one hot encoding.

```python
x = pd.get_dummies(X , drop_first = True)
```

Train Test Split

The dataset is divided into 80 percent training data and 20percent testing data. The training data will be used to train the Random Forest model while the testing data will be used to evaluate its performance on unseen data.

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 42)
```

Train Random Forest Regressor

model combines multiple decision trees and averages their predictions to improve performance..avg = mean

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(n_estimators = 100 , random_state = 42)
```

```python
rf_model.fit(x_train, y_train)
```

```python
y_pred = rf_model.predict(x_test)
```

 Model Evaluation

We compare the actual ratings with the predicted ratings using MAE, MSE, RMSE  R² to measure the performance of the Random Forest model.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
```

```python
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("R-squared:", r2)
```

# ** Principal Component Analysis (PCA)**

PCA is a dimensionality reduction technique used to reduce the number of features while retaining as much useful information as possible.

 PCA will be used to reduce the dimensionality of the player feature data before applying XGBoost.

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(x)
```

```python
pca.fit(X_scaled)
```

```python
pca.explained_variance_ratio_
```

```python
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
```

```python
n_compenents = 0
```

```python
for i , value in enumerate(cumulative_variance):
    if value >= 0.95:
        print(i + 1)
        break
```

```python
pca_22 = PCA(n_components = 22)
```

```python
x_pca = pca_22.fit_transform(X_scaled)
```

```python
x_pca.shape
```

PCA Result

PCA reduced the original feature space from 41 features to 22 principal components while retaining approximately 95% of the variance

```python
x_train_pca , X_test_pca , y_train , y_test = train_test_split(x_pca , y , test_size = 0.2 , random_state = 42)
```

# **XGBoost Regression**

After reducing the feature dimensions using PCA.... XGBoost Regression is applied to predict the player's overall rating.

```python
from xgboost import XGBRegressor
```

```python
xgb_model = XGBRegressor(n_estimators = 100 , random_state = 42)
```

```python
xgb_model.fit(x_train_pca , y_train)
```

```python
xgb_pred = xgb_model.predict(X_test_pca)
```

```python
xg_mae = mean_absolute_error(y_test , xgb_pred)
xg_mse = mean_squared_error(y_test , xgb_pred)
xg_rmse = mean_squared_error(y_test , xgb_pred )**0.5
xg_r2 = r2_score(y_test , xgb_pred)
```

```python
print("Mean Absolute Error:", xg_mae)
print("Mean Squared Error:", xg_mse)
print("Root Mean Squared Error:", xg_rmse)
print("R-squared:", xg_r2)
```

```python
pc_names = [f"PC{i}" for i in range(1,23)]
```

```python
xgb_importance = pd.Series( xgb_model.feature_importances_ , index = pc_names)
```
